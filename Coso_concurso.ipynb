```python
class Concurso:
    """
    Clase Concurso
    - Atributo: mensaje
    - Métodos: codificar() y decodificar()
    - Vocal: se mueve 2 a la derecha (U → E)
    - Vocal con tilde: se mueve 2 a la izquierda (Ó → É)
    - Consonante: se mueve 1 a la izquierda (D → C)
    - Ejemplo: GAKI → HOLA
    """
    def __init__(self, mensaje):
        self.mensaje = mensaje.upper()
        self.vocales = "AEIOU"
        self.vocales_tilde = "ÁÉÍÓÚ"
        self.alfabeto = "ABCDEFGHIJKLMNÑOPQRSTUVWXYZ"

    def codificar(self):
        resultado = ""
        for letra in self.mensaje:
            if letra in self.vocales:
                pos = self.vocales.index(letra)
                nueva = self.vocales[(pos + 2) % len(self.vocales)]
                resultado += nueva
            elif letra in self.vocales_tilde:
                pos = self.vocales_tilde.index(letra)
                nueva = self.vocales_tilde[(pos - 2) % len(self.vocales_tilde)]
                resultado += nueva
            elif letra in self.alfabeto:
                pos = self.alfabeto.index(letra)
                nueva = self.alfabeto[(pos - 1) % len(self.alfabeto)]
                resultado += nueva
            else:
                resultado += letra
        return resultado

    def decodificar(self):
        resultado = ""
        for letra in self.mensaje:
            if letra in self.vocales:
                pos = self.vocales.index(letra)
                nueva = self.vocales[(pos - 2) % len(self.vocales)]
                resultado += nueva
            elif letra in self.vocales_tilde:
                pos = self.vocales_tilde.index(letra)
                nueva = self.vocales_tilde[(pos + 2) % len(self.vocales_tilde)]
                resultado += nueva
            elif letra in self.alfabeto:
                pos = self.alfabeto.index(letra)
                nueva = self.alfabeto[(pos + 1) % len(self.alfabeto)]
                resultado += nueva
            else:
                resultado += letra
        return resultado



mensaje = input("Escribe el mensaje: ").upper()
accion = input("¿Quieres codificar o decodificar? ").lower()

c = Concurso(mensaje)

if accion == "codificar":
    print("Mensaje codificado:", c.codificar())
elif accion == "decodificar":
    print("Mensaje decodificado:", c.decodificar())
else:
    print("Opción no válida.")
```
