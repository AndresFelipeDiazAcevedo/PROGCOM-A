```python
def salto_derecha(a):
  for i in range(a):
    print("salta a la derecha")
```

```python
def balancear_brazos_arriba_abajo(a, b):
    for i in range(max(a, b)):
        if i < a:
            print("brazo derecho, sube y baja")
        if i < b:
            print("brazo izquierdo, sube y baja")
```

```python
balancear_brazos_arriba_abajo(2,1)
```
