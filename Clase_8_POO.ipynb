```python
# Floristeria

#1. CONJUNTO DE FLORES, se elige aleatoriamente una cantidad
import random

flores=["tulipan","lirio","rosa","dahlia","margarita","girasol","clavel","geranio"]
seleccion=random.sample(flores,3)
seleccion
```

```python
def inventario():
  flores=["tulipan","lirio","rosa","dahlia","margarita","girasol","clavel","geranio"]
  seleccion=random.sample(flores,3)

  inventario={
      flor:{
          "precio":random.randint(2000,6000),
          "cantidad":random.randint(10,50),
          "vendidas":0
      } for flor in seleccion
  }
  return inventario
```

```python
inventario()
```

```python
def menu(inventario):
  print("\n---MENÚ DE FLORES DISPONIBLES FLORISTERIA COSO ---")
  for flor, datos in inventario.items():
    print(f"{flor}:${datos['precio']} canti {datos['cantidad']} disponible")
  print("--------------------------------")
```

```python
menu(inventario())
```

```python
def ventas(inventario,flor,cantidad,total_ganado):
  if flor not in inventario:
    print(f"{flor} no se encuentra a la venta ")
    return total_ganado
  datos=inventario[flor]
  if cantidad>datos["cantidad"]:
    print(f"lo siento no tenemos la cantidad requerida")
    return total_ganado

  #Realizar la venta
  datos["cantidad"]-=cantidad
  datos["vendidas"]+=cantidad
  venta = datos["precio"]*cantidad
  total_ganado+=venta
  print(f"Vendidas:{cantidad} de {flor} por un total de ${venta}")

  #Si se acaban, eliminamos la flor
  if datos["cantidad"]==0:
    print(f"{flor} ha sido elminada del inventario")
    del inventario[flor]
  return total_ganado
```

```python
def cerrar_tienda(inventario_inicial, inventario_final, total_ganado):
  print("\n ---- Tienda Cerrada ----")
  #Combinar el inventario incial con el final para saber cuantas se vendio
  flores_vendidas={}
  for flor, datos in inventario_inicial.items():
    cantidad_final=inventario_final.get(flor,{"cantidad":0})["cantidad"]
    vendidas=datos["cantidad"]-cantidad_final
    flores_vendidas[flor]=vendidas

  #Determinar cuales fueron las mas vendidas
    mas_vendida=max(flores_vendidas,key=flores_vendidas.get)
    menos_vendida=min(flores_vendidas,key=flores_vendidas.get)
    print(f"La flor mas vendida fue {mas_vendida}:{flores_vendidas[mas_vendida]} unidad(es)")
    print(f"La flor menos vendida fue {menos_vendida}:{flores_vendidas[menos_vendida]} unidad(es)")
```
