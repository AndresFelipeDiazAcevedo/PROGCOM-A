#lista por comprension
forma nueva de generar lista a partir de una expresion
sintaxis:


```
nombre_lista=[expresion for item in iterable if condicion]
```
1. expresion es el valor que generara la nueva lista
2. item: variable que representa cada elemento dentro de una iteracion
3. iterable: cuanto elementos va a tener la nueva lista
4. condicion: es opcional y se emplea si los elementos deben cumplir algo



````python
#convertir lista de oraciones en lista de palabras
oraciones=["1 oracion IHSIDA", "2 oracionn eawanf", "3 oracion apde"]
palabras=[i.split() for i in oraciones]
palabras
````

````python
#clasificar si un numero es positivo o negativo

numeros=[1,4,6,-20-2-69,4]
lista2=[(n,"positivo") if n>0 else (n,"negativo") for n in numeros]
lista2
````

````python
temperaturas = [float(input(f"Temperatura {i+1}: ")) for i in range(10)]

resultado = ["Frío" if t < 20 else "Templado" if t <= 27 else "Caliente" for t in temperaturas]

print(resultado)
````
