##3. con parametros, sin retorno

###4. con parametros, con retorno

````python
#EJEMPLO: area de un rectangulo
def area_rectangulo(u,du):
  return f"El area de un rectangulo de base {u} y altura {du} es {u*du}"
````

````python
a=area_rectangulo(10,15)
````

````python
print(a)
````

##Funciones anonimas o lambda
1. Siempre usan parametros
2. Siempre dan retorno
3. no tienen nombre --> anonimas
4. **Se puede emplear n parametros, pero solo ejecuta 1 linea de codigo**

### Definicion

```
variable=lambda (p,p1,p2...pn):operacion
```



````python
#calcular area triangulo
area_triangulo=lambda a,b: a*b/2
````

````python
area_triangulo(15,10)
````

````python
area_circulo=lambda a: (3.14)*a**2
````

````python
area_circulo(3)
````

````python
edad_mayor=lambda a,b,c: max(a,b,c)
````

````python
edad_mayor(7,8,10)
````

````python
edad_mayor_terniario=lambda a,b,c: a if a>b and a>c else b if  b>c else c
````

````python
edad_mayor_terniario(15,17,2)
````

````python
#6
def contar_vocales(texto):
    return " tiene", sum(1 for c in texto.lower() if c in "aeiou"), "vocales"
````

````python
#10
def nombres_vocal(lista):
    return sum(1 for nombre in lista if nombre[0].lower() in "aeiou")
````

````python
print(nombres_vocal(["andres", "adrian", "juan", "david", "asas", "juliana", "carlos"]))
````

````python
nombres_vocal_ = lambda lista: sum(1 for nombre in lista if nombre[0].lower() in "aeiou")
````

````python
print(nombres_vocal_(["andres", "adrian", "juan", "david", "asas", "juliana", "carlos"]))
````
