Un auto parte del reposo y acelera uniformemente con
a=2 𝑚/s2 durante t=10s

Calcular:

Velocidad final

Distancia recorrida

```python
# Datos iniciales
v0 = 0.0
a = 2.0
t = 10.0

# Cálculos con fórmulas de MRUA
vf = v0 + a * t
x = v0 * t + 0.5 * a * t**2


print("Velocidad final:", vf, "m/s")
print("Distancia recorrida:", x, "m")
```
