```python
import random

def inventario():
    flores = ["tulipan", "lirio", "rosa", "dahlia", "margarita", "girasol", "clavel", "geranio"]
    seleccion = random.sample(flores, 3)

    inventario = {
        flor: {
            "precio": random.randint(2000, 6000),
            "cantidad": random.randint(10, 50)
        }
        for flor in seleccion
    }
    return inventario

print("Bienvenido a la Floristeria COSO COSO")

opcion = input("\n¿Desea comprar una flor? (si/no): ").lower()

if opcion != "si":
    print("\nVuelva pronto")
else:
    stock = inventario()
    compras = []
    total_general = 0

    seguir = "si"
    while seguir == "si":
        print("\nFlores disponibles:")
        for flor, datos in stock.items():
            print(f"- {flor}: {datos['cantidad']} unidades, ${datos['precio']} ")

        tipo = input("\n¿Cuál flor desea comprar?: ").lower()

        if tipo in stock:
            cantidad = int(input("¿Cuántas unidades desea?: "))

            if cantidad <= stock[tipo]["cantidad"]:
                subtotal = cantidad * stock[tipo]["precio"]
                total_general += subtotal
                stock[tipo]["cantidad"] -= cantidad

                compras.append({
                    "flor": tipo,
                    "cantidad": cantidad,
                    "precio": stock[tipo]["precio"],
                    "subtotal": subtotal
                })

                print(f"\n Compra realizada. Subtotal: ${subtotal}")
            else:
                print("\n No hay suficientes unidades disponibles.")
        else:
            print("\n Esa flor no está disponible.")

        seguir = input("\n¿Desea comprar otra flor? (si/no): ").lower()


    print("\n--------- FACTURA----------")
    if compras:
        for item in compras:
            print(f"{item['cantidad']} x {item['flor']} (${item['precio']} c/u) = ${item['subtotal']}")
        print(f"\nTOTAL A PAGAR: ${total_general}")
    else:
        print("no se realizo ninguna compra")

    print("\nCompra terminada")
```
